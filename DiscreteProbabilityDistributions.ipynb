```python
#Bernolli
#[SPECIAL CASE OF BINOMIAL WITH ONE TRIAL]
#x is an element of {0,1}
#mean=p variance=p

def bernoulli_distribution_function_probability_calculator(x,p):
    q=1-p
    return (p**x)*(q**(1-x))

x=int(input('Please enter x: '))    
p=float(input('Please enter p: '))
print(f'The probability of x={x} and p={p} is: {bernoulli_distribution_function_probability_calculator(x,p):.4f}')
q=1-p
variance=p*q
print(f'The mean of this distribution is: {p:.10f}')
print(f'The variance of this distribution is: {var:.10f}')
```

```python
#Binomial
#x is an element of {0,1,2,3,...,n}
#mean=np variance=npq
from math import *

def binomial_distribution_function_probability_calculator(n,x,p):
    q=1-p
    return comb(n,x)*(p**x)*(q**(n-x))
    
n=int(input('Please enter n: '))
x=int(input('Please enter x: '))
p=float(input('Please enter p: '))
q=1-p
mean=n*p
variance=n*p*q
print(f'The probability of n={n} x={x} and p={p} is: {binomial_distribution_function_probability_calculator(n,x,p):.10f}')
print(f'The mean of this distribution is: {mean:.10f}')
print(f'The variance of this distribution is: {variance:.10f}')
```

```python
#Negative binomial distribution
#mean=xq/p variance=xq/p^2
#x>0,n=x,x+1,x+2,x+3,...
from math import *

def negative_binomial_distribution_function_probability_calculator(n,x,p):
    q=1-p
    return comb(n-1,x-1)*(p**x)*(q**(n-x))

n=int(input('Please enter n: '))
x=int(input('Please enter x: '))
p=float(input('Please enter p: '))
q=1-p
mean=x*q/p
variance=x*q/(p**2)
print(f'The probability of n={n} x={x} and p={p} is: {negative_binomial_distribution_function_probability_calculator(n,x,p):.10f}')
print(f'The mean of this distribution is: {mean:.10f}')
print(f'The variance of this distribution is: {variance:.10f}')
```

```python
#Geometric distribution
#[SPECIAL CASE OF NEGATIVE BINOMIAL DISTRIBUTION WITH X=1]
#x=1,n=1,2,3,4,...

def geometric_distribution_function_probability_calculator(n,p):
    q=1-p
    return p*(q**(n-1))

n=int(input('Please enter n: '))
p=float(input('Please enter p: '))
q=1-p
mean=q/p
variance=q/p**2

print(f'The probability of n={n} x={1} and p={p} is: {geometric_distribution_function_probability_calculator(n,p):.10f}')
print(f'The mean of this distribution is: {mean:.10f}')
print(f'The variance of this distribution is: {variance:.10f}')
```

```python
# Hyper geometric distribution
#Trials are not independent
#samples selected without replacement
#probability of success or failure keep on changing
from math import *

def hypergeometric_distribution_function_probability_calculator(N,n,k,x):
    return comb(k,x)*comb(N-k,n-x)/comb(N,n)

N=int(input('Please enter N: '))
n=int(input('Please enter n: '))
k=int(input('Please enter k: '))
x=int(input('Please enter x: '))
mean=n*k/N
p=k/N
q=1-p
variance=n*p*q*((N-n)/(N-1))
print(f'The mean of this distribution is: {mean:.10f}')
print(f'The variance of this distribution is: {variance:.10f}')

probability_holder=0
for i in range(3):
    probability_holder+=hypergeometric_distribution_function_probability_calculator(N,n,k,i)

print(f'The probability of getting atmost 2 indivuduals with high blood pressure when choosing {n}\
 individuals is: {probability_holder:.10f}')
```

```python
#Poisson distribution
#men=variance=λ x=0,1,2,3,4,5,...
from math import *

def poisson_distribution_function_probability_calculator(labda,x):
    return (labda**x)*exp(-labda)/factorial(x)

labda=float(input('please input λ: '))
x=int(input('Please input x: '))
print(f'The probability of λ={labda} and x={x} is: {poisson_distribution_function_probability_calculator(labda,x):.10f}')
mean=labda
variance=labda
print(f'The mean is {labda}')
print(f'The variance is {labda}')
```
