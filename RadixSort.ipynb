```python
def counting_sort_for_radix(arr, exp):
    n = len(arr)

    output = [0] * n
    count = [0] * 10  # digit range: 0–9

    # Count occurrences of each digit at 'exp'
    for num in arr:
        index = (num // exp) % 10
        count[index] += 1

    # Convert count[] so it contains actual positions
    for i in range(1, 10):
        count[i] += count[i - 1]

    # Build the output array (iterate backwards for stability)
    for i in range(n - 1, -1, -1):
        index = (arr[i] // exp) % 10
        output[count[index] - 1] = arr[i]
        count[index] -= 1

    # Copy back to original array
    for i in range(n):
        arr[i] = output[i]


def radix_sort(arr):
    # Find max number to know the number of digits
    max_num = max(arr)

    exp = 1  # start with units place

    while max_num // exp > 0:
        counting_sort_for_radix(arr, exp)
        print(f"After sorting by digit place {exp}: {arr}")
        exp *= 10


# Example usage:
arr = [564, 213, 987, 432, 123, 765, 321, 654, 876]
print("Original array:", arr)
radix_sort(arr)
print("Final sorted array:", arr)
```

The order of the radix sort is from least significant digit to most significant digit as displayed in the program above.
The time complexity of radix sort is O(d*(n+k)). d is the number of digits in the largest number; n is the number of elements in the list and k is the range of each digit (typically 0-9 for decimal numbers)
