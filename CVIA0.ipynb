```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# Define input dimension
input_dim = 20  # Example: number of input features

# Define the model
class Net(nn.Module):
    def __init__(self, input_dim):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(input_dim, 64)
        self.fc2 = nn.Linear(64, 10)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# Instantiate the model
model = Net(input_dim)

# Test with a random input
x = torch.randn(1, input_dim)  # batch size 1
output = model(x)
print(output)
```
