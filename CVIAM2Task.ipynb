# **Import Libraries**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

# **Upload and Load a Color Image**

```python
# Upload an image
uploaded = files.upload()

# Get the uploaded filename
filename = next(iter(uploaded))

# Read the image using OpenCV
image_bgr = cv2.imread(filename)

# Check that the image was loaded successfully
if image_bgr is None:
    raise ValueError("Image could not be loaded. Please upload a valid JPG or PNG image.")

# Convert BGR to RGB for correct display with Matplotlib
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

# Display the original image
plt.figure(figsize=(8, 6))
plt.imshow(image_rgb)
plt.title("Original RGB Image")
plt.axis("off")
plt.show()
```

# **Display the RGB Color Channels**

```python
# Split the RGB image into individual channels
red, green, blue = cv2.split(image_rgb)

# Display the three RGB channels
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(red, cmap='gray')
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green, cmap='gray')
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue, cmap='gray')
plt.title("Blue Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

**Interpretation**

Each channel is displayed as a grayscale image:

**Red channel:** Shows the intensity of red at each pixel. Bright areas contain more red.
**Green channel:** Shows the intensity of green. Bright areas contain more green.
**Blue channel:** Shows the intensity of blue. Bright areas contain more blue.

The three channels together form the complete RGB color image.

# **Convert the Image to HSV**

```python
# Convert RGB image to HSV
image_hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)

# Split HSV channels
hue, saturation, value = cv2.split(image_hsv)
```

# **Display the HSV Channels**

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(hue, cmap='gray')
plt.title("Hue Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(saturation, cmap='gray')
plt.title("Saturation Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(value, cmap='gray')
plt.title("Value Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

**Interpretation**
**Hue:** Represents the basic color, such as red, green, blue, or yellow. In OpenCV, Hue normally ranges from 0–179 for 8-bit images.
**Saturation:** Represents the intensity or purity of the color. Low saturation produces grayish or dull colors, while high saturation produces vivid colors.
**Value:** Represents brightness. Higher values indicate brighter pixels, while lower values indicate darker pixels.

# **Display HSV Channels Using Appropriate Ranges**

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(hue, cmap='hsv')
plt.title("Hue")
plt.colorbar()
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(saturation, cmap='gray', vmin=0, vmax=255)
plt.title("Saturation")
plt.colorbar()
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(value, cmap='gray', vmin=0, vmax=255)
plt.title("Value")
plt.colorbar()
plt.axis("off")

plt.tight_layout()
plt.show()
```

# **Why may HSV be more suitable than RGB for certain computer vision tasks?**

HSV can be more suitable than RGB for some computer vision tasks because it separates color information (Hue) from color intensity (Saturation) and brightness (Value). In RGB, the red, green, and blue components are all affected by changes in illumination, making it more difficult to identify an object's color consistently when lighting conditions change.

For example, consider an object that is red. Under bright illumination, its RGB values may be relatively high, while under poor lighting its RGB values may become much lower. Although the object is still red, its RGB representation has changed significantly. In HSV, the Hue component can remain relatively stable because it represents the dominant color independently of brightness to a greater extent.

This makes HSV useful for tasks such as object detection, color-based segmentation, tracking, and image processing. For example, a computer vision system can identify pixels corresponding to a particular range of hues even when the brightness of the scene varies.

However, HSV is not completely invariant to lighting conditions. Strong shadows, highlights, changes in illumination color, and very low saturation can still affect the HSV values. Therefore, HSV should be viewed as a more convenient representation for separating color from brightness rather than as a perfect solution to color constancy.

# **Conclusion**

HSV provides a useful alternative to RGB for computer vision applications where identifying and separating colors is important. RGB directly represents colors using red, green, and blue intensities, whereas HSV separates color characteristics into Hue, Saturation, and Value. This separation makes HSV particularly useful for color-based segmentation and object detection because algorithms can focus primarily on Hue and Saturation while treating brightness separately. Consequently, HSV can provide more robust results than RGB in situations involving moderate changes in lighting, although significant illumination changes can still affect both representations.
